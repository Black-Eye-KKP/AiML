```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([[0.5, 1.5], [1, 1], [1.5, 1.5], [2, 2], [2.5, 2.5]])
y = np.array([0, 0, 0, 1, 1])
```

```python
fig, ax = plt.subplots(1, 1, figsize=(5, 5))
ax.set_xlim(0, 4)
ax.set_ylim(0, 3.5)

# Plot the two classes with different colors
ax.scatter(x[y == 0, 0], x[y == 0, 1], c='blue', label='Class 0')
ax.scatter(x[y == 1, 0], x[y == 1, 1], c='red', label='Class 1')

# Decision boundary: x1 + x2 = 3
x_vals = np.linspace(0, 4, 100)
y_vals = 3 - x_vals
ax.plot(x_vals, y_vals, '--', c='black', linewidth=2, label='Decision boundary')

ax.set_ylabel('y')
ax.set_xlabel('x')
ax.set_title('Decision Boundary')
ax.legend()
plt.show()
```

```python
z = np.arange(-10, 20 )
fig,ax = plt.subplots(1, 1, figsize=(5, 5))
g = 1/(1 + np.exp(-z))
ax.plot(z, g, linewidth=2)
ax.set_title("Sigmoid Function")
ax.set_xlabel("z")
ax.set_ylabel("g(z)")
ax.axvline(0, color="black", linewidth=2, linestyle="--")
plt.show()
```
