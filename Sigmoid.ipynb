```python
def sigmoid(z):
  g = 1 / (1 + np.exp(-z))
  return g
```

```python
z_tmp =np.arange(-10,11)
print(z_tmp)
y = sigmoid(z_tmp)
np.set_printoptions(precision=3) #for **(.***)
print(np.c_[z_tmp, y]) #side wise column stacking np.c_
```

```python
fig,ax = plt.subplots(1,1,figsize=(5,3))
ax.plot(z_tmp,y, c="b")
ax.set_title("Sigmoid Function")
ax.set_xlabel("z")
ax.set_ylabel("g(z)")
ax.axline((0, 0.5), slope=0, color="black", linewidth=2, linestyle="--")
```
