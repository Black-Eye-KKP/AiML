```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_train = np.array([[0.5,1.0],[0.9,1.2],[1.0,1.5],[1.5,1.8],[2.5,2.5],[4.0,2.9]])
y_train = np.array([0,0,0,1,1,1])
```

```python
fig,ax = plt.subplots(1,1,figsize=(5,5))
ax.scatter(x_train[:,0],x_train[:,1],c=y_train)
x_val = np.linspace(0, 4, 100)
y_val = 3 - x_val
ax.plot(x_val, y_val, color="black", linewidth=2, linestyle="--")
ax.set_title("Training Data")
ax.set_xlabel("Feature 1")
ax.set_ylabel("Feature 2")
legend = ax.legend(["Class 0","Class 1"],loc="lower left")
plt.show()
```

```python
def sigmoid(z):
    return 1/(1 + np.exp(-z))
def compute_cost(x, y, w, b):
    m = x.shape[0]
    cost = 0.0
    for i in range(m):
        z = np.dot(x[i], w) + b
        f_wb = sigmoid(z)
        cost += y[i]*np.log(f_wb) + (1-y[i])*np.log(1-f_wb)
    cost = -cost/m
    return cost
```

```python
w_tmp = np.array([1.0, 1.0])
b_tmp = -3.0
print(compute_cost(x_train, y_train, w_tmp, b_tmp))
```
